```python
x = 5
y = "John"

print(x)
print(y)
```

#Variable Names
A variable can have a short name (like x and y) or a more descriptive name (age, carname, total_volume).

Rules for Python variables:

* A variable name must start with a letter or the underscore character

* A variable name cannot start with a number

* A variable name can only contain alpha-numeric characters and
underscores (A-z, 0-9, and _ )

* Variable names are case-sensitive (age, Age and AGE are three different variables)

* A variable name cannot be any of the Python keywords.

```python
# no Namespces while declaring variable name

first name = "Rawat"

print(first name)
```

# Formatting String

F-String was introduced in Python 3.6, and is now the preferred way of formatting strings.

To specify a string as an f-string, simply put an f in front of the string literal, and add curly brackets {} as placeholders for variables and other operations.

```python
age = 36

txt = f"My name is John, I am {age}"

print(txt)
```

#String Methods

String methods return new values. They do not change the original string.

```python
# 1. upper() – Convert to uppercase

name = "bhupender"

print(name.upper())
```

```python
# 2. lower() – Convert to lowercase

name = "BHUPENDER"

print(name.lower())
```

```python
# 3. capitalize() – Capitalize first character

text = "python programming"

print(text.capitalize())
```

```python
# 4. title() – Capitalize every word

text = "python data analytics"

print(text.title())
```

```python
# 5. strip() – Remove spaces from beginning and end

text = "   Python   "

print(text.strip())
```

```python
# 6. replace() – Replace text

text = "I like Excel"

result = text.replace("Excel", "Power BI")

print(result)
```

```python
# 7. split() – Convert string into a list

text = "Python,SQL,Power BI"

result = text.split(",")

print(result)
```

```python
# 8. join() – Join list elements into a string

skills = ["Python", "SQL", "Power BI"]

result = ", ".join(skills)

print(result)
```

```python
# 9. find() – Find position of text

text = "I am learning Python"

print(text.find("Python"))
```

```python
# 10. count() – Count occurrences

text = "Python is easy. Python is powerful."

print(text.count("Python"))
```

```python
# 11. startswith() – Check beginning

text = "Python Programming"

print(text.startswith("Python"))
```

```python
# 12. endswith() – Check ending

file = "sales_report.xlsx"

print(file.endswith(".xlsx"))
```

```python
# 13. isalpha() – Check whether all characters are alphabets

name = "Bhupender"

print(name.isalpha())
```

```python
# 14. isdigit() – Check whether all characters are digits

age = "35"

print(age.isdigit())
```
